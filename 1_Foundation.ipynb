## What is a generative AI (GenAI) system?
When you think about the word AI, what comes to mind? Is it Terminators walking through a barren wasteland, a white robot with a blue hue, or is it ChatGPT? A more formal definition states that GenAI is a computational technique capable of generating "new" content from training data. This new content could be in many forms, such as audio, video, images, text, etc. Given this definition, we can claim that ChatGPT is a generative AI system.

GenAI systems (i.e., systems using generative AI models as a main component) can be classified into five main types:
- text generation; examples include ChatGPT, Llama app, Claude
- image generation: text to image and vice versa; examples include DALL-E, Stable Diffusion, MidJourney
- Audio/Music Generation: text to sound (audio, music, speech) and vice versa; examples include Suno, MusicLM
- Code Generation: text to code; examples include Github Copilot, Claude, Vertex AI
- speech generation: speech to speech and speech to video; examples include Sora, ResembleAI

These GenAI systems often involve the use of a large language model (LLM), which can require many resources.

### What impact does the creation of a GenAI system have on the environment?
Every GenAI system has come at a cost, mainly due to the GenAI model. For example, it was said that training simple model emitted about the same amount of carbon as a trans-American flight.

These models are powered by data centers: specialized buildings equipped with computers with powerful chips needed to run the large models we use in our daily lives. These centers get extremely hot, powering all the technological activity we do, and requiring cooling systems to regulate their internal temperatures, typically using water [daxsom, 2024]. GenAI systems are particularly resource-intensive and need corresponding data centers to operate. According to the International Energy Agency, the sum of centers, over 10,000 and 30% in the United States, will contribute 6% of total electricity consumption by 2026 [International Energy Agency, 2024].

## What we mean by Critical AI Literacy (CAIL)
Critical AI literacy is the skills and knowledge needed to develop and utilize AI technologies with a critical mindset [Velander, 2024]. CAIL caters to increasing the awareness of AI’s double-edged sword. However, CAIL helps individuals increase productivity with work processes, fosters a deeper understanding of AI bias, and promotes fair and diverse assistance to all types of communities. Some specific CAIL benefits to AI technology include portraying diverse backgrounds, undermining traditionally marginalized voices in designs, highlighting the environment, and understanding repercussions from over-dependence on GenAI systems [Veldhuis, 2024].

The introduction of ChatGPT has brought both good and bad on students and teachers alike. Teachers have begun using these systems to create schedule plans and rubrics while having to deal with increased plagiarism [Huang, 2024]. Those with a lackluster level of CAIL will not be able to pinpoint that chatbots conflate information with absolute truth and disseminate harmful information targeted at historically marginalized groups [Tanksley, 2024].1

## Basic NLP Terminology
- Tokenization is the process of converting words into numerical tokens
- Afterward, transformers are used to look at the entire sentence before predicting the next word. The self-attention module determines which words are important 

```python
%pip install sentencepiece accelerate

model_path = "allenai/Olmo-3-7B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_path, use_fast=False, trust_remote_code=True)

model = AutoModelForCausalLM.from_pretrained(
    model_path,
    device_map="auto",
    torch_dtype='auto'
).eval()
```

Did you notice the model path name above? We have chosen Map-Neo as our model of choice instead of others like GPT's and Llama's models, primarily for three reasons:
- open-source over proprietary: Open-source models offer transparency, which allows for high understandability versus the black-box OpenAI models. 
- Customizability: utilizing an open-source model enables adaptive customization. We can modify the architecture of our system using fine-tuning, retrieval-augmented generation (RAG), and adding advanced modules from high-level machine learning methodologies. 
- The open access to the dataset is a key feature of Map-NEO. The researchers claim it to be the largest, most diverse, and transparent pre-training corpus. Unlike OpenAI, this dataset
is open to everyone to research how the data can influence outputs.

>Think about the previous cost mentioned when discussing models. How much would the cost differ from an open-source and proprietary model?

## Generating a response from our generative model

```python
input_text = "A long, long time ago,"

input_ids = tokenizer(input_text, add_generation_prompt=True, return_tensors='pt').to(model.device)
output_ids = model.generate(**input_ids, max_new_tokens=20)

response = tokenizer.decode(output_ids[0], skip_special_tokens=True)
print(response)
```

## AI Agent Setup

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

class GenoVault:
    def __init__(self, model_path):
        self.model_path = model_path
        
    def generate(self, token_amount, input):
        tokenizer = AutoTokenizer.from_pretrained(self.model_path, use_fast=False, trust_remote_code=True)

        model = AutoModelForCausalLM.from_pretrained(
            self.model_path,
            device_map="auto",
            torch_dtype='auto'
        ).eval()
        
        input_ids = tokenizer(input, add_generation_prompt=True, return_tensors='pt').to(model.device)
        output_ids = model.generate(**input_ids, max_new_tokens=token_amount)
        response = tokenizer.decode(output_ids[0], skip_special_tokens=True)
        return response

gv = GenoVault(model_path="allenai/Olmo-3-7B-Instruct")
```

## What does multiprompting do?
Lots of prompting = lots of token requests

```python
class GenoVault:
    def __init__(self, model_path):
        self.model_path = model_path
        
    def generate(self, input, history, tokens):
        tokenizer = AutoTokenizer.from_pretrained(self.model_path, use_fast=False, trust_remote_code=True)

        model = AutoModelForCausalLM.from_pretrained(
            self.model_path,
            device_map="auto",
            torch_dtype='auto'
        ).eval()
        
        input_ids = tokenizer(input, add_generation_prompt=True, return_tensors='pt').to(model.device)
        
        if tokens == 0: 
            tokens = 100
        
        output_ids = model.generate(**input_ids, max_new_tokens=tokens)
        response = tokenizer.decode(output_ids[0], skip_special_tokens=True)
        
        for i in range(len(response)):
            time.sleep(0.02)
            yield response[:i+1]

gv = GenoVault(model_path="allenai/Olmo-3-7B-Instruct")

demo = gr.ChatInterface(
    gv.generate,
    type="messages",
    additional_inputs=[
        gr.Slider(100, 300),
    ],
    textbox=gr.Textbox(placeholder="What would you like to talk about?", container=False, scale=7),
    flagging_mode="manual",
    flagging_options=["Like", "Spam", "Inappropriate", "Other"],
    save_history=True,
)

demo.launch()  
```

If you get any interesting responses, press the flag button and it'll record it. Might be useful if you want to analyze why the model responded in such a way or if you want to cross check an output. Things flagged will appear in .gradio\flagged\dataset1.csv
